## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
# Ranked content actions + reason codes
actions = [
    {"rank": 1, "action": "Refresh or update high-priority content", "reason_code": "STALE_HIGH_PRIORITY"},
    {"rank": 2, "action": "Review content with declining performance", "reason_code": "PERFORMANCE_DROP"},
    {"rank": 3, "action": "Review content with strong recent engagement", "reason_code": "HIGH_ENGAGEMENT"},
    {"rank": 4, "action": "Monitor low-confidence recommendations", "reason_code": "LOW_CONFIDENCE"},
]

for item in actions:
    print(f"{item['rank']}. {item['action']} [{item['reason_code']}]")
```

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

```python
print("INTENDED USE:")
print("- Prioritize content for human review and refresh.")
print("- Support editors with ranked, evidence-based recommendations.")
print("- Use model scores as decision support, not as final decisions.")

print("\nLIMITS:")
print("- Predictions are not guarantees of future performance.")
print("- Recommendations may be affected by data quality, freshness, and distribution shift.")
print("- Human review is required before consequential content decisions.")
```

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

```python
print("HUMAN REVIEW:")
print("- Review high-priority recommendations before publishing or removing content.")
print("- Check the evidence, reason code, and context before taking action.")
print("- Escalate uncertain or high-impact cases to a human reviewer.")

print("\nNO-GO LIST:")
print("- Do not automatically publish, delete, or demote content.")
print("- Do not make decisions using protected or sensitive personal information.")
print("- Do not treat model predictions as guaranteed outcomes.")
```

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

```python
print("MONITORING:")
print("- Track precision, ranking quality, coverage, and recommendation outcomes.")
print("- Monitor data freshness and changes in feature distributions.")
print("- Review false positives and false negatives regularly.")

print("\nRETRAIN TRIGGERS:")
print("- Significant performance degradation.")
print("- Major distribution or data drift.")
print("- Important changes in user or content behavior.")
print("- Sufficient new labeled data becomes available.")
```

## 5. Exports for the paper

*Write the queue (and any figures you want to reuse) to work/outputs/ — your paper builds on these files.*

```python
import os
import json
import pandas as pd

os.makedirs("../outputs", exist_ok=True)
os.makedirs("../figures", exist_ok=True)

queue = pd.DataFrame(actions)
queue.to_csv("../outputs/ranked_action_queue.csv", index=False)

metrics = {
    "purpose": "Content action prioritization",
    "human_review_required": True,
    "automation": False
}

with open("../outputs/action_playbook_metrics.json", "w") as f:
    json.dump(metrics, f, indent=2)

print("Exported:")
print("../outputs/ranked_action_queue.csv")
print("../outputs/action_playbook_metrics.json")
```
